```python
import ee
import geemap

# 1. Autenticación e inicialización de Google Earth Engine
ee.Authenticate()
ee.Initialize(project='ee-vshalisko')

# 2. Cargar la colección WorldClim V1
worldclim = ee.ImageCollection('WORLDCLIM/V1/MONTHLY')

# 3. Filtrar o seleccionar la imagen que contiene las bandas mensuales
# WorldClim V1 almacena todas las variables en una sola imagen de la colección
dataset = worldclim.select(['prec'])

# 4. Seleccionar precipitacion dutante la temporada seca y humeda
precip_seca = dataset.filter(ee.Filter.gte('month', 3).And(ee.Filter.lte('month', 5)))
precip_humeda = dataset.filter(ee.Filter.gte('month', 8).And(ee.Filter.lte('month', 10)))

# 5. Calcular suma de precipitación para los periodos
precip_seca_sums = precip_seca.reduce(ee.Reducer.sum()).rename('precip_seca')
precip_humeda_sums = precip_humeda.reduce(ee.Reducer.sum()).rename('precip_humeda')
```

```python
# 6. Definir parámetros de visualización para la precipitación (en mm)
vis_params = {
    'min': 0,
    'max': 1500,
    'palette': ['blue', 'limegreen', 'yellow', 'orange', 'red']
}

# 7. Crear el mapa interactivo y agregar la capa
Map = geemap.Map(center=[19.43, -99.13], zoom=4, basemap='OpenTopoMap') # Centrado por defecto
Map.addLayer(precip_seca_sums, vis_params, 'P seca (mm)')
Map.addLayer(precip_humeda_sums, vis_params, 'P húmeda (mm)')
Map.add_colorbar(vis_params, label="Precipitación (mm)")

# Mostrar mapa en Colab
Map
```
